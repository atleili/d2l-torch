### 数据操作
#### 关于张量

```python
# 导入pytorch库
import torch
```

张量时一个拥有多个维度的数组，具有一个维度的张量对应数学中的向量，具有两个维度的张量对应数学中的矩阵
使用arange可以创建一个行向量

```python
# 创建一个行向量，元素从0开始，步长为1，长度为12
x = torch.arange(12)
x
```

通过shape属性可以得到张量的形状

```python
x.shape
```

通过numel()方法可以得到张量中元素的数量

```python
x.numel()
```

通过reshape()方法可以改变张量的形状

```python
X = x.reshape(3,4)
print(X)
X.shape
```

对于使用reshape()方法修改张量形状，不需要手动指定每个维度的大小
例如对于需要一个二维张量，指定其中一个维度大小，另一个维度写入-1，pytorch会自动计算另一个维度的大小

```python
Y = x.reshape(2, -1)
print(Y)
Y.shape
```

通过下面的方法可以分别创建一个指定形状的全0、全1、符合某个特定概率分布的张量

```python
# 创建一个指定形状的全0张量
x = torch.zeros(2, 3, 4)
x = torch.zeros((2,3,4))
# 上面两个方法都是可行的，第一种方式分别指定每个维度的大小，第二种方式传入了一个shape元组

# 创建一个指定形状的全1张量
x = torch.ones(2, 3, 4)

# 创建一个服从标准高斯分布的随机采样张量
x = torch.randn(3, 4)
```

通过下面的方式可以创建一个有确定赋值的张量

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
x
```
