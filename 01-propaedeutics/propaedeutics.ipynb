通过下面的方法可以分别创建一个指定形状的全0、全1、符合某个特定概率分布的张量

#### 运算符
torch张量支持逐元素计算和矩阵运算两种方式，如下：

```python
# 逐元素计算
x = torch.tensor([1, 2, 3, 4])
y = torch.tensor([5, 6, 7, 8])
z = x + y # 加
print(z)
z = x - y     # 减
print(z)
z = x * y     # 乘
print(z)
z = x / y     # 除
print(z)
z = x // y    # 整除，向下取整
print(z)
z = x ** y    # 幂
print(z)
z = x % y     # 取模
# 逐元素计算方法
z = torch.exp(x) # 取自然指数
print(z)

# 矩阵运算
x = torch.arange(12).reshape(4, 3)
print(x)
y = torch.tensor([[2, 0, 4, 3], [1, 5, 6, 1], [1, 0, 0, 5]]) 
z = x @ y # 矩阵乘法
print(z)
# 矩阵转置
z = x.T
print(z)
z = x.t() # 相比于T，t()方法会返回一个新张量，而T会返回一个视图，直接修改T返回的张量会修改原张量，而直接修改t()返回的张量不会修改原张量(参见后面原地操作部分)

# sum()方法可以对张量元素求和，返回一个单元素张量
z.sum()
```

通过cat方法，可以把多个张量在一个指定维度拼接(concatenate)在一起

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
y = torch.tensor([[7, 8, 9], [10, 11, 12]])
print(torch.cat((x, y), dim=0))
print(torch.cat((x, y), dim=1))
```

#### 索引和切片

```python
x = torch.arange(20).reshape(4, 5)
print(x)
# 使用[-1]访问最后一个元素
print(x[0, -1])
# 使用[a:b]访问张量中第a个到第b(不包含)个元素的切片
print(x[1, 1:3])
# 使用[a:b:c]访问张量中第a个到第b(不包含)个元素，步长为c的切片
print(x[1, 1:4:2])
# 使用[:]访问所有元素
print(x[:, 1])
```

#### 原地操作
对于一些操作，如x = x + y，一般来说，七过程为，开辟一块内存存储计算x + y得到的值，然后把x指向该值，在大量计算时，就涉及了频繁内存分配，对此，使用id()函数可以验证该情况，id()函数提供了内存中引用对象的确切地址。

```python
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])
before = id(x)
x = x + y
after = id(x)
print(before == after)
```


通过一些简单的操作，可以在原始内存上进行修改，以避免该问题。
pytorch专门为这些操作提供了后缀为_的方法，如add_() sub_()等，这些方法会直接在原始内存上进行修改，而不会开辟新的内存空间。

```python
x = torch.tensor([1, 2, 3], dtype=torch.float32)
y = torch.tensor([4, 5, 6], dtype=torch.float32)

x.add_(y)     # 元素加
print(x)
x.sub_(y)     # 元素减
print(x)
x.mul_(y)     # 元素乘
print(x)
x.div_(y)     # 元素除
print(x)
x.floor_divide_(y) # 元素向下整除
print(x)
x.remainder_(y) # 元素取模
print(x)
x.pow_(y) # 元素幂
print(x)
x.relu_() # 元素relu激活
print(x)
x.sigmoid_() # 元素sigmoid激活
print(x)
x.tanh_() # 元素tanh激活
print(x)
x.clamp_(1, 2) # 元素取值范围限制(min, max)
print(x)
x.copy_(y) # 复制
print(x)
```

这些操作中部分对应的自运算操作符也是原地操作，如下：

```python
x += y
print(x)
x -= y
print(x)
x *= y
print(x)
x /= y
print(x)
x //= y
print(x)
x %= y
print(x)
x **= y
print(x)
```

切片操作是一种创建视图的操作，而不是创建新的张量，因此，切片操作不会开辟新的内存空间，利用这一特点也可以减少内存开销：

```python
before = id(x)
x[:] = x + y
print(id(x))
print(before == id(x))
```

上述操作先创建了一个临时张量保存x + y的结果，然后回头赋值给x。
值得注意的是，由于切片操作的这种特性，在使用切片的时候需要注意一些问题防止错误修改了原本张量，例如：

```python
print(x)
y = x[:]
y.add_(x)
print(x)
y.add(x)
print(x)
```

上面的示例中，张量y通过切片操作创建了一个x的视图，此时，使用原地操作修改y，实际上x的值也被修改了，
而使用非原地操作修改y，x的值并没有被修改，因此，在使用切片操作时，需要格外小心，避免错误修改了原本张量。这点对于 += -= 等原地操作符号同样适用。
矩阵的转置中，直接使用属性x.T会返回一个视图，修改该视图也会修改原张量

```python
print(x)
y = x.T
y[0] = 100
print(x)
```

### 数据预处理
#### 读取数据集
首先创建`../data/house_tiny.csv`文件，写入数据以创建一个简单人工数据集

```python
import os
os.makedirs('../../data', exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w') as f:
  f.write('NumRooms,Alley,Price\n') # 列名
  f.write('NA,Pave,127500\n') # 每行表示一个数据样本
  f.write('2,NA,106000\n')
  f.write('4,NA,178100\n')
  f.write('NA,NA,140000\n')
```

采用pandas包中的read_csv函数读取数据集

```python
import pandas as pd
data = pd.read_csv(data_file)
print(data)
```

#### 处理缺失值
数据中NaN表示缺失值，一般可以通过插值法或者删除法处理，插值法通过特定插值方式处理，删除法直接忽略对应数据
这里使用插值法处理

```python
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]
inputs = pd.get_dummies(inputs, dummy_na=True)
inputs = inputs.fillna(inputs.mean())
print(inputs)
```

#### 转换为张量
在上述处理后，所有数据都转换为数值类型了(其中部分布尔类型也可以作为数字形式正常计算)
通过转换成张量形式可以方便后续操作

```python
X = torch.tensor(inputs.to_numpy(dtype=float))
Y = torch.tensor(outputs.to_numpy(dtype=float))
print(X)
print(Y)
```
